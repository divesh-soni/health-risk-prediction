# Conclusion (diabetes)

#### Glucose, BloodPressure, SkinThickness, Insulin, BMI -> 0 values are actually missing. Insulin has 374 zeros out of 768 (almost half), SkinThickness 227
#### fill them with median (Insulin and SkinThickness are right skewed with outliers so median is better than mean)
#### median only from train data, otherwise test data leaks into training
#### Outcome is imbalanced (~65% no / 35% yes) -> use stratify in split and check precision recall also
#### Glucose has highest correlation with Outcome, then BMI, Age, Pregnancies
